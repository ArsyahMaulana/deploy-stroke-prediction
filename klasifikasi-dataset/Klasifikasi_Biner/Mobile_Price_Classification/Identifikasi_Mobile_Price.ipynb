```python
import glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
```

## 2. Data Loading

```python
DATA_PATH = "train.csv"   # file train.csv dari dataset Mobile Price Classification
df = pd.read_csv(DATA_PATH)
print("Ukuran data (baris, kolom):", df.shape)
df.head()
```

## 3. Identifikasi Jenis Klasifikasi

Notebook ini **tidak** membuat model. Tujuannya membuktikan lewat data bahwa dataset ini bukan **Klasifikasi Biner**.

```python
TARGET = "price_range"
print("Jumlah kelas target:", df[TARGET].nunique())
print(df[TARGET].value_counts().sort_index())
sns.countplot(x=df[TARGET]); plt.title("Distribusi price_range"); plt.show()
```

## 4. Kesimpulan


        Kolom target `price_range` punya **4 kelas** (0, 1, 2, 3), masing-masing sama banyak. Tiap HP hanya punya satu label, tetapi pilihan label-nya lebih dari 2, jadi dataset ini **multikelas**, bukan biner.
