## 2. Data Loading

```python
FILES = sorted(glob.glob("*.csv") + glob.glob("*.xlsx"))
print("File data di folder ini:", FILES)
assert FILES, "Taruh file dataset (.csv / .xlsx) di folder yang sama dengan notebook ini"
DATA_PATH = FILES[0]   # ganti manual kalau ada lebih dari satu file
df = pd.read_excel(DATA_PATH) if DATA_PATH.endswith(".xlsx") else pd.read_csv(DATA_PATH)
print("Ukuran data (baris, kolom):", df.shape)
df.head()
```

## 3. Identifikasi Jenis Klasifikasi

Notebook ini **tidak** membuat model. Tujuannya membuktikan lewat data bahwa dataset ini bukan **Klasifikasi Multikelas**.

```python
# Kolom label = kolom berhuruf kapital tunggal (A, B, C, ...). Cek hasilnya, sesuaikan jika perlu.
LABEL_COLS = [c for c in df.columns if len(str(c)) == 1 and str(c).isupper()]
print("Kolom label:", LABEL_COLS)

jumlah_label_aktif = df[LABEL_COLS].sum(axis=1)
print("\nBanyak artikel menurut jumlah label aktif:")
print(jumlah_label_aktif.value_counts().sort_index())
print(f"\nArtikel dengan lebih dari 1 label aktif: {(jumlah_label_aktif > 1).mean() * 100:.2f}%")
```

## 4. Kesimpulan


        Satu artikel bisa memiliki **beberapa label aktif sekaligus**. Pada multikelas, tiap baris hanya boleh punya **satu** kelas. Karena itu dataset ini **multilabel**, bukan multikelas.
