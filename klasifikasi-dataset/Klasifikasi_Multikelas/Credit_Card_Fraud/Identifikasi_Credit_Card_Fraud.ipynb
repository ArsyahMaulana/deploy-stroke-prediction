```python
import glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
```

## 2. Data Loading

```python
DATA_PATH = "creditcard.csv"
df = pd.read_csv(DATA_PATH)
print("Ukuran data (baris, kolom):", df.shape)
df.head()
```

## 3. Identifikasi Jenis Klasifikasi

Notebook ini **tidak** membuat model. Tujuannya membuktikan lewat data bahwa dataset ini bukan **Klasifikasi Multikelas**.

```python
TARGET = "Class"
print("Jumlah kelas target:", df[TARGET].nunique())
print(df[TARGET].value_counts())
print(f"\nPersentase fraud: {df[TARGET].mean() * 100:.3f}%")
df[TARGET].value_counts().plot(kind="bar", logy=True)
plt.title("Distribusi Class (skala log)"); plt.xticks(rotation=0); plt.show()
```

## 4. Kesimpulan


        Kolom target `Class` hanya punya **2 nilai** (0 dan 1), jadi dataset ini **biner**, bukan multikelas.
        Dataset ini juga sangat tidak seimbang (fraud hanya sebagian kecil dari seluruh transaksi), tetapi jumlah kelasnya tetap dua.
