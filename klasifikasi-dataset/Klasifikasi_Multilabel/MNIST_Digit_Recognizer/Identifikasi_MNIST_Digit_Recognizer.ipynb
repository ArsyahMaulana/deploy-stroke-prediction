```python
import glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
```

## 2. Data Loading

```python
DATA_PATH = "train.csv"   # train.csv dari kompetisi Digit Recognizer
df = pd.read_csv(DATA_PATH)
print("Ukuran data (baris, kolom):", df.shape)
df.iloc[:, :8].head()
```

## 3. Identifikasi Jenis Klasifikasi

Notebook ini **tidak** membuat model. Tujuannya membuktikan lewat data bahwa dataset ini bukan **Klasifikasi Multilabel**.

```python
TARGET = "label"
print("Jumlah kelas target:", df[TARGET].nunique())
print(df[TARGET].value_counts().sort_index())
print("\nJumlah kolom label per baris: 1 (kolom 'label'), sisanya adalah piksel gambar")
sns.countplot(x=df[TARGET]); plt.title("Distribusi digit"); plt.show()
```

## 4. Kesimpulan


        Setiap gambar hanya punya **satu** digit (0 sampai 9), jadi ada 10 kelas yang saling lepas dan tiap baris hanya memiliki satu label. Itu ciri **multikelas**, bukan multilabel.
