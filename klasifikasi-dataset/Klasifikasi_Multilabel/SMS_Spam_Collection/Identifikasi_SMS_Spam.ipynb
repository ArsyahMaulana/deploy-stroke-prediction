```python
import glob, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
```

## 2. Data Loading

```python
DATA_PATH = "spam.csv"
df = pd.read_csv(DATA_PATH, encoding="latin-1")
df = df.iloc[:, :2]                 # hanya dua kolom pertama (label, teks)
df.columns = ["label", "text"]
print("Ukuran data (baris, kolom):", df.shape)
df.head()
```

## 3. Identifikasi Jenis Klasifikasi

Notebook ini **tidak** membuat model. Tujuannya membuktikan lewat data bahwa dataset ini bukan **Klasifikasi Multilabel**.

```python
TARGET = "label"
print("Jumlah kelas target:", df[TARGET].nunique())
print(df[TARGET].value_counts())
sns.countplot(x=df[TARGET]); plt.title("Distribusi label"); plt.show()
```

## 4. Kesimpulan


        Hanya ada **satu** kolom target dengan **2 nilai** (`ham` dan `spam`). Satu SMS tidak mungkin ham dan spam sekaligus, jadi dataset ini **biner**, bukan multilabel.
